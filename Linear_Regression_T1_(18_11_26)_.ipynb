```python
import numpy as np
import matplotlib.pyplot as plt

# Ensure variables are available (fallback if previous cell was not executed)
try:
    _ = slope
    _ = intercept
    _ = x
    _ = y
except NameError:
    x = np.array([1, 2, 3, 4, 5])
    y = np.array([3, 6, 9, 12, 15])
    x_mean = np.mean(x)
    y_mean = np.mean(y)
    slope = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean) ** 2)
    intercept = y_mean - slope * x_mean

# Calculate the predicted y values using the regression equation
y_pred = slope * x + intercept

# Plot the actual data points and the regression line
plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='blue', label='Data Points (Actual)')
plt.plot(x, y_pred, color='red', linewidth=2, label=f'Regression Line: y = {slope:.2f}x + {intercept:.2f}')

# Add labels, title and legend
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Linear Regression Visualization')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

x = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # 2D array venum
y = np.array([3, 6, 9, 12, 15])

model = LinearRegression()
model.fit(x, y)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```
