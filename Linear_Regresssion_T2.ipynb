```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Define variables
X = df[['Study_Hours']]
y = df['Exam_Marks']

# Initialize and fit the model
model = LinearRegression()
model.fit(X, y)

# Get metrics
r_sq = model.score(X, y)
slope = model.coef_[0]
intercept = model.intercept_

print(f"Coefficient (Slope): {slope:.4f}")
print(f"Intercept: {intercept:.4f}")
print(f"R-squared Score: {r_sq:.4f}")
```

```python

# ============================================================
# LINEAR REGRESSION - STUDENT PERFORMANCE PREDICTION
# ============================================================

# STEP 1: IMPORT REQUIRED LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import warnings
warnings.filterwarnings('ignore')


# STEP 2: UPLOAD DATASET

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_excel(file_name)

print("Dataset Loaded Successfully!")
print("\nFirst Five Records:")
print(df.head())


# STEP 3: DATASET INFORMATION

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
print(df.info())

print("\nStatistical Description:")
print(df.describe())


# STEP 4: CHECK MISSING VALUES

print("\nMissing Values:")
print(df.isnull().sum())

# Handle missing numerical values
df = df.fillna(df.median(numeric_only=True))

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())


# STEP 5: CHECK DUPLICATE VALUES

print("\nDuplicate Records:")
print(df.duplicated().sum())

df = df.drop_duplicates()


# STEP 6: DATA VISUALIZATION

# Histogram
df.hist(figsize=(12, 8))
plt.suptitle("Distribution of Student Data")
plt.tight_layout()
plt.show()


# STEP 7: CORRELATION ANALYSIS

plt.figure(figsize=(10, 6))

sns.heatmap(
    df.drop(columns=['Student_ID']).corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap")
plt.show()


# STEP 8: SIMPLE LINEAR REGRESSION

# Independent Variable (X)
# Study_Hours

# Dependent Variable (Y)
# Exam_Marks

X = df[['Study_Hours']]
y = df['Exam_Marks']


# STEP 9: SPLIT DATASET INTO TRAINING AND TESTING

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("\nTraining Data Size:", X_train.shape)
print("Testing Data Size:", X_test.shape)


# STEP 10: CREATE LINEAR REGRESSION MODEL

model = LinearRegression()


# STEP 11: TRAIN THE MODEL

model.fit(X_train, y_train)

print("\nModel Training Completed Successfully!")


# STEP 12: DISPLAY MODEL COEFFICIENTS

print("\nIntercept:", model.intercept_)

print("Coefficient:", model.coef_[0])


# STEP 13: LINEAR REGRESSION EQUATION

print("\nLinear Regression Equation:")

print(
    "Exam_Marks =",
    round(model.intercept_, 2),
    "+",
    round(model.coef_[0], 2),
    "* Study_Hours"
)


# STEP 14: PREDICT TEST DATA

y_pred = model.predict(X_test)

print("\nActual and Predicted Values:")

result = pd.DataFrame({
    'Actual Marks': y_test.values,
    'Predicted Marks': np.round(y_pred, 2)
})

print(result)


# STEP 15: MODEL EVALUATION

mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

n = len(y_test)
p = X_test.shape[1]

adjusted_r2 = (
    1 - (1 - r2) * (n - 1) / (n - p - 1)
    if n > p + 1 else np.nan
)

print("\n========== MODEL PERFORMANCE ==========")

print("Mean Absolute Error (MAE):", round(mae, 2))

print("Mean Squared Error (MSE):", round(mse, 2))

print("Root Mean Squared Error (RMSE):", round(rmse, 2))

print("R2 Score:", round(r2, 4))

print("Adjusted R2 Score:", round(adjusted_r2, 4))


# STEP 16: REGRESSION LINE VISUALIZATION

plt.figure(figsize=(8, 5))

plt.scatter(
    X_test['Study_Hours'],
    y_test,
    color='blue',
    label='Actual Values'
)

plt.plot(
    X_test['Study_Hours'],
    y_pred,
    color='red',
    label='Regression Line'
)

plt.xlabel("Study Hours")
plt.ylabel("Exam Marks")

plt.title("Simple Linear Regression")

plt.legend()
plt.grid(True)
plt.show()


# STEP 17: ACTUAL VS PREDICTED GRAPH

plt.figure(figsize=(8, 5))

plt.scatter(y_test, y_pred, color='green')

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    color='red'
)

plt.xlabel("Actual Marks")
plt.ylabel("Predicted Marks")

plt.title("Actual vs Predicted Values")

plt.show()


# STEP 18: RESIDUAL CALCULATION

residuals = y_test - y_pred

print("\nResidual Values:")
print(residuals.values)


# STEP 19: RESIDUAL PLOT

plt.figure(figsize=(8, 5))

plt.scatter(y_pred, residuals)

plt.axhline(
    y=0,
    color='red',
    linestyle='--'
)

plt.xlabel("Predicted Marks")
plt.ylabel("Residuals")

plt.title("Residual Plot")

plt.show()


# STEP 20: RESIDUAL DISTRIBUTION

plt.figure(figsize=(8, 5))

sns.histplot(
    residuals,
    kde=True
)

plt.title("Residual Distribution")

plt.xlabel("Residual Values")

plt.show()


# STEP 21: PREDICT NEW STUDENT MARKS

new_student = pd.DataFrame({
    'Study_Hours': [5, 8, 10, 12]
})

new_predictions = model.predict(new_student)

print("\n========== NEW STUDENT PREDICTIONS ==========")

for hours, marks in zip(
    new_student['Study_Hours'],
    new_predictions
):
    print(
        "Study Hours:", hours,
        "Predicted Exam Marks:", round(marks, 2)
    )


# ============================================================
# STEP 22: MULTIPLE LINEAR REGRESSION
# ============================================================

# Select Multiple Independent Variables

X_multi = df[
    [
        'Study_Hours',
        'Attendance_Percent',
        'Previous_Marks',
        'Assignment_Score'
    ]
]

y_multi = df['Exam_Marks']


# Split Dataset

X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(
    X_multi,
    y_multi,
    test_size=0.20,
    random_state=42
)


# Create Multiple Linear Regression Model

multi_model = LinearRegression()


# Train Model

multi_model.fit(X_train_m, y_train_m)

print("\nMultiple Linear Regression Training Completed!")


# Display Coefficients

print("\nMultiple Regression Intercept:")
print(multi_model.intercept_)

print("\nMultiple Regression Coefficients:")

for feature, coefficient in zip(
    X_multi.columns,
    multi_model.coef_
):
    print(feature, ":", round(coefficient, 4))


# Predict Multiple Regression Values

y_multi_pred = multi_model.predict(X_test_m)


# Evaluate Multiple Linear Regression

multi_mae = mean_absolute_error(
    y_test_m, y_multi_pred
)

multi_mse = mean_squared_error(
    y_test_m, y_multi_pred
)

multi_rmse = np.sqrt(multi_mse)

multi_r2 = r2_score(
    y_test_m, y_multi_pred
)


print("\n========== MULTIPLE LINEAR REGRESSION ==========")

print("MAE:", round(multi_mae, 2))

print("MSE:", round(multi_mse, 2))

print("RMSE:", round(multi_rmse, 2))

print("R2 Score:", round(multi_r2, 4))


# STEP 23: COMPARE BOTH MODELS

comparison = pd.DataFrame({
    'Evaluation Metrics': [
        'MAE',
        'MSE',
        'RMSE',
        'R2 Score'
    ],

    'Simple Linear Regression': [
        mae,
        mse,
        rmse,
        r2
    ],

    'Multiple Linear Regression': [
        multi_mae,
        multi_mse,
        multi_rmse,
        multi_r2
    ]
})

print("\n========== MODEL COMPARISON ==========")

print(comparison.round(4))


# STEP 24: MODEL COMPARISON GRAPH

comparison_plot = comparison.set_index(
    'Evaluation Metrics'
)

comparison_plot.plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title("Simple vs Multiple Linear Regression")

plt.ylabel("Metric Values")

plt.xticks(rotation=0)

plt.legend()

plt.grid(axis='y')

plt.show()


# STEP 25: PREDICT USING MULTIPLE REGRESSION

new_student_data = pd.DataFrame({

    'Study_Hours': [6],

    'Attendance_Percent': [85],

    'Previous_Marks': [65],

    'Assignment_Score': [75]

})

final_prediction = multi_model.predict(
    new_student_data
)

print("\n========== FINAL STUDENT PERFORMANCE ==========")

print(
    "Predicted Exam Marks:",
    round(final_prediction[0], 2)
)


print("\n====================================")
print("LINEAR REGRESSION PROGRAM COMPLETED")
print("====================================")
```
